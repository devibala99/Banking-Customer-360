## 01 - Data Discovery

Purpose: Inspect the raw banking dataset, identify available tables, understand their size, and prepare for data relationship and quality analysis.

## 1. Load and Inspect the Raw Dataset

Load the Excel workbook and identify all available tables.

```python
import pandas as pd

file_path = "../data/raw/banking_customer360_synthetic.xlsx"

excel_file = pd.ExcelFile(file_path)
excel_file.sheet_names

sheets = pd.read_excel(
    file_path,
    sheet_name = None
)
sheets.keys()

for sheet_name, df in sheets.items():
    print(f"{sheet_name}: {df.shape}")
```

### Initial Findings

The dataset contains six related tables:
Customers, Accounts, Transactions, Loans, Customer_Service, and Data_Dictionary.

Next, we will inspect the Data Dictionary and identify the key fields and relationships between these tables.

## 3. Table Structure

Review the columns in each table to identify customer IDs, account IDs, transaction fields, loan fields, and service-related fields.

```python
for sheet_name, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(f"{sheet_name}")
    print(f"{'=' * 60}")
    print(df.columns.tolist())
```

## 4. Data Types

Review the data types of each column to identify numeric, text, date, and other fields that may require cleaning or transformation.

```python
for sheet_name, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(f"{sheet_name}")
    print(f"{'=' * 60}")
    print(df.dtypes)
```

## 5. Missing Values

Check the number of missing values in each table as part of the initial data quality assessment.

```python
for sheet_name, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(f"{sheet_name}")
    print(f"{'=' * 60}")
    print(df.isnull().sum())
```

## 6. Duplicate Records

Check for duplicate rows in each table before data cleaning.

```python
for sheet_name, df in sheets.items():
    print(f"{sheet_name}: {df.duplicated().sum()} duplicate rows")
```

## 7. Sample Records

Review a small sample of each table to understand the actual values stored in the dataset.

```python
for sheet_name, df in sheets.items():
    print(f"\n{'=' * 60}")
    print(f"{sheet_name}")
    print(f"{'=' * 60}")
    display(df.head(3))
```

## 8. Identify Key Fields

Inspect columns containing ID fields to identify possible primary keys and foreign keys for the Customer 360 data model.

```python
for sheet_name, df in sheets.items():
    id_columns = [
        col for col in df.columns
        if "id" in col.lower()
    ]
    
    print(f"{sheet_name}: {id_columns}")
```

## 9. Key Uniqueness Check

Check whether candidate ID fields contain unique values. This helps identify primary keys and fields used to connect related tables.

```python
for sheet_name, df in sheets.items():
    id_columns = [
        col for col in df.columns
        if "id" in col.lower()
    ]
    
    print(f"\n{sheet_name}")
    
    for col in id_columns:
        print(
            f"{col}: "
            f"{df[col].nunique()} unique / "
            f"{len(df)} total"
        )
```

## 10. Table Relationships

Compare ID fields across tables to identify how customers, accounts, transactions, loans, and service interactions are connected.

```python
id_values = {}

for sheet_name, df in sheets.items():
    id_values[sheet_name] = {
        col: set(df[col].dropna().unique())
        for col in df.columns
        if "id" in col.lower()
    }

for sheet_name, columns in id_values.items():
    print(f"\n{'=' * 60}")
    print(sheet_name)
    print(f"{'=' * 60}")
    
    for col, values in columns.items():
        print(f"{col}: {len(values)} unique values")
```

## 11. Data Discovery Summary

Create a compact summary of the structure and data quality of all tables.

```python
discovery_summary = []

for sheet_name, df in sheets.items():
    discovery_summary.append({
        "Table": sheet_name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Duplicate_Rows": df.duplicated().sum(),
        "Missing_Values": df.isnull().sum().sum()
    })

discovery_summary = pd.DataFrame(discovery_summary)

discovery_summary
```

## 12. Data Discovery Findings

The raw dataset contains six related tables covering customers, accounts, transactions, loans, customer service interactions, and field definitions.

The initial inspection covers table size, columns, data types, missing values, duplicate records, candidate ID fields, and possible relationships.

The next stage will use these findings to define the data model and determine how the tables should be cleaned and integrated into a Customer 360 dataset.
